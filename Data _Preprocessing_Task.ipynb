## MinMax Scalar

```python
from sklearn.preprocessing import MinMaxScaler

mm_scaler = MinMaxScaler()
df_mm = df.copy()
df_mm[num_col] = mm_scaler.fit_transform(df[num_col])
```

## Plot the scaled distribution

## Robust Scalar

```python
df_rb[num_col].hist(figsize=(12,10), bins=20)
plt.suptitle("Robust Scaled Distribution")
plt.show()
```

1. How did scaling reshape the distribution?

-Scaling does not change the shape of the distribution (skewness, peaks), but it changes the scale and spread of the data. StandardScaler recenters data around zero, MinMaxScaler compresses values into a fixed range [0,1], and RobustScaler adjusts data based on median and IQR, reducing the influence of extreme values.

2. Which scaler handled outliers most effectively?

RobustScaler handled outliers most effectively because it uses the median and interquartile range instead of mean and standard deviation. This prevents extreme values from distorting the scaling process.

3. How MinMaxScaler affects feature ranges

MinMaxScaler transforms all values into a fixed range [0,1], preserving the original distribution shape. However, the presence of outliers causes most data points to be compressed into a narrow range, reducing the effectiveness of feature representation.

4. How StandardScaler affects mean and variance

StandardScaler standardizes data by transforming it to have a mean of 0 and a standard deviation of 1. While this centers the data, it remains sensitive to outliers, which can still skew the distribution.
