## Task 2

```python
X = df.drop('job_satisfaction_score', axis=1)
```

## Task 3: Identify Feature Types


```python
num_col=df.select_dtypes(include="number").columns
print(num_col)
```

```python
cat_col=df.select_dtypes(include="object").columns
print(cat_col)
```

```python
categorical_cols = X.select_dtypes(include=['object', 'category']).columns
print(categorical_cols)
```

```python
binary_cols = []
nominal_cols = []

for col in categorical_cols:
    unique_vals = X[col].nunique()
    
    if unique_vals == 2:
        binary_cols.append(col)
    else:
        nominal_cols.append(col)

print("Binary categorical:", binary_cols)
print("Nominal categorical:", nominal_cols)
```

## Task 4: Encode Categorical Variables




```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

binary_cols = ['gender', 'uses_focus_apps', 'has_digital_wellbeing_enabled']

for col in binary_cols:
    X[col] = le.fit_transform(X[col])
```

```python
X = pd.get_dummies(
    X,
    columns=['job_type', 'social_platform_preference'],
    drop_first=True
)
```
