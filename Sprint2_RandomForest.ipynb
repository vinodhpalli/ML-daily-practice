## • Column structure


### 1.Make
#### Car brand strongly affects MSRP because premium brands like BMW, Mercedes-Benz, and Audi usually have higher prices compared to economy brands like Maruti Suzuki or Hyundai. Brand reputation, reliability, and luxury positioning directly increase vehicle price.


### 2.Model
#### Different models within the same company have different prices because of features, performance, and design. For example, SUV or sports models generally cost more than entry-level hatchbacks.


### 3.Year
#### Newer manufacturing years usually increase MSRP because modern cars include updated technology, better safety systems, and improved fuel efficiency. Older models tend to have lower prices.


### 4.Engine Fuel Type
#### Fuel type affects pricing based on efficiency and technology. Hybrid and electric vehicles are usually more expensive because of advanced battery systems, while diesel and petrol cars may have moderate pricing differences.


### 5.Engine HP
#### Higher horsepower generally increases MSRP because powerful engines provide better acceleration and performance. Sports and luxury cars usually have very high HP and higher prices.


### Engine Cylinders
#### Cars with more cylinders (6, 8, or 12 cylinders) are usually more expensive because they produce greater power and smoother performance. Small economy cars mostly use 3 or 4 cylinders with lower prices.


### 6.Transmission Type
#### Automatic transmission vehicles are often priced higher than manual cars because they provide easier driving experience and advanced transmission systems.


### 7.Driven Wheels
#### AWD (All-Wheel Drive) and 4WD vehicles usually have higher MSRP than FWD cars because they offer better traction, off-road capability, and improved driving stability.


### 8.Number of Doors
#### Vehicles with more doors generally indicate family-oriented or premium vehicles, which may slightly increase MSRP, though this feature alone has limited impact compared to engine or brand features.


### 9.Market Category
#### Luxury, performance, crossover, and hybrid categories usually have higher MSRP because they target premium customers and include advanced features. Economy market segments have lower prices.


### 10.Vehicle Size
#### Larger vehicles like SUVs and full-size sedans are usually more expensive because they require larger engines, more materials, and offer more passenger space. Compact cars tend to be cheaper.


### 11.Vehicle Style
#### Sports cars, convertibles, and SUVs often have higher MSRP than hatchbacks or sedans because of design, performance, and market demand.


### 14.Highway MPG
#### Vehicles with better highway mileage may have slightly higher MSRP if they use advanced fuel-saving technologies. However, extremely high-performance cars often sacrifice mileage for power.


### 15.City MPG
#### Better city mileage increases value for fuel-efficient vehicles, especially hybrids and compact cars. Fuel-efficient cars are often priced higher due to advanced engine technology.


### 16.Popularity
#### Highly popular vehicles may have higher MSRP because of strong market demand and brand recognition. However, popularity alone does not guarantee a high price.


### 17.MSRP (Target Column)
#### MSRP represents the manufacturer’s suggested retail price of the vehicle. It is influenced by engine performance, brand value, technology, fuel efficiency, luxury features, safety systems, and market demand.

# Step 5: Outlier Detection & Treatment


```python
plt.figure(figsize=(20,10))

df.boxplot(rot=90)

plt.title("Outliers Detection for Total DataFrame")
plt.xlabel("Columns")
plt.ylabel("Values")

plt.show()
```

#### Columns Like Engine HP,Highway MPG,City MPG,MSRP Contain High Outliers
#### Columns Like Year,Engine Cylinder,Number of Doors,Popularity contain Low Outliers

```python
import numpy as np

# Select numeric columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Capping outliers
    df[col] = np.where(df[col] < lower_bound,
                       lower_bound,
                       df[col])

    df[col] = np.where(df[col] > upper_bound,
                       upper_bound,
                       df[col])

print("Outliers treated successfully")
```

```python
plt.figure(figsize=(20,10))

df.boxplot(rot=90)

plt.title("Outliers Detection for Total DataFrame")
plt.xlabel("Columns")
plt.ylabel("Values")

plt.show()
```

# Step 6: Feature Encoding &Step 7: Feature Scaling

```python
x = df.drop("MSRP (Manufacturer's suggested retail Price)", axis=1)

y = df["MSRP (Manufacturer's suggested retail Price)"]
```

# Step 8: Train-Test Split¶

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
num_col=x_train.select_dtypes(include='number').columns
cat_col=x_train.select_dtypes(include='object').columns
num_col
cat_col
```

```python
x.info()
```

```python
cat_col
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
import pandas as pd

# Numerical columns
num_col = x.select_dtypes(include=['int64', 'float64']).columns

# Ordinal columns
ordinal_col = ['Vehicle Size']

# One Hot Encoding columns
onehot_col = ['Make','Model','Engine Fuel Type','Transmission Type','Driven_Wheels','Market Category','Vehicle Style']


ct = ColumnTransformer(transformers=[("oe",OrdinalEncoder(
         categories=[['Compact', 'Midsize', 'Large']],
         handle_unknown='use_encoded_value',
         unknown_value=-1),
     ordinal_col),

    ("ohe",
     OneHotEncoder(handle_unknown='ignore'),
     onehot_col),

    ("sc",
     StandardScaler(),
     num_col)

])

# Train data transform
x_train_transform = ct.fit_transform(x_train)

# Convert to DataFrame
x_train_transform = pd.DataFrame(
    x_train_transform.toarray() if hasattr(x_train_transform, "toarray") else x_train_transform
)

print(x_train_transform.head())

# Test data transform
x_test_transform = ct.transform(x_test)

# Convert to DataFrame
x_test_transform = pd.DataFrame(
    x_test_transform.toarray() if hasattr(x_test_transform, "toarray") else x_test_transform
)

print(x_test_transform.head())
```

# SPRINT 2: MODEL BUILDING & EVALUATION


###  Step 1: Baseline Model


```python
from sklearn.ensemble import RandomForestRegressor

# Create model
model = RandomForestRegressor(random_state=42)

# Train model
model.fit(x_train_transform, y_train)
```

####  Accuracy

```python
y_pred = model.predict(x_test_transform)
```

```python
from sklearn.metrics import r2_score

accuracy = r2_score(y_test, y_pred) * 100

print("Accuracy :", accuracy)
```

##### The Linear Regression model achieved an accuracy (R² Score) of 96.25%, indicating excellent predictive performance on the test dataset.

##  Step 2: Model Evaluation


### Regression Metrics:


#### MAE Code

```python
# Predict on test data
y_pred = model.predict(x_test_transform)

# MAE
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("MAE :", mae)
```

### MAE (Mean Absolute Error)

###### MAE: The model achieved an MAE of 2383.47, indicating that the predicted car prices differ from the actual prices by about 2383 on average.

#### RMSE Code

```python
from sklearn.metrics import mean_squared_error

rmse = mean_squared_error(y_test, y_pred) ** 0.5

print("RMSE :", rmse)
```

### RMSE (Root Mean Squared Error)

###### The model achieved an RMSE of 4025.77, showing the average prediction error magnitude with higher penalty for large errors.

#### R² Score Code

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R2 Score :", r2)
```

### R² Score (Coefficient of Determination)

###### The model achieved an R² Score of 95.90%, indicating excellent predictive performance and strong model fit on the test dataset.

## Step 3: Overfitting & Underfitting Check


#### Step 1: Training Predictions

```python
y_train_pred = model.predict(x_train_transform)
y_train_pred
```

#### Step 2: Testing Predictions

```python
y_test_pred = model.predict(x_test_transform)
y_test_pred
```

#### Step 3: Calculate R² Scores

```python
from sklearn.metrics import r2_score

train_score = r2_score(y_train, y_train_pred)

test_score = r2_score(y_test, y_test_pred)

print("Training R2 Score :", train_score)

print("Testing R2 Score  :", test_score)
```

#####  The Decision Tree Regression model achieved a Training R² Score of 0.9910 and a Testing R² Score of 0.9589. Since both scores are high and the difference between them is small, the model does not show significant overfitting or underfitting and generalizes well on unseen data.

###  Hyperparameter Tuning


```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor

# Parameters
params = {
    'n_estimators': [50, 100],
    'max_depth': [5, 10, 15],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}

# GridSearchCV
grid = GridSearchCV(
    RandomForestRegressor(random_state=42),
    param_grid=params,
    cv=5,
    scoring='r2',
    n_jobs=-1
)
grid.fit(x_train_transform, y_train)

# Best parameters
print("Best Parameters:", grid.best_params_)

# Best CV score
print("Best CV Score:", grid.best_score_)
```

##### The Decision Tree Regressor performed strongly with an R² score of 96.4%, and the best model used a depth of 15 with controlled splitting to balance accuracy and overfitting.

```python
best_model = grid.best_estimator_

# Train score
train_score = best_model.score(x_train_transform, y_train)

# Test score
test_score = best_model.score(x_test_transform, y_test)

print("Train Score:", train_score)
print("Test Score:", test_score)
```
