```python
## target variable quilty category
y=df["quality_category"]
```

```python
## input varibles 13 columns
x=df.drop(["quality_category"],axis=1)
```

```python
x.shape
```

```python
## apply column transformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder,StandardScaler
ct=ColumnTransformer(transformers=[("oe",OrdinalEncoder(),[0,1,4]),
                                   ("sc",StandardScaler(),[2,3,5,6,7,8,9,10,11,12])])
```

```python
x_transformed=ct.fit_transform(x)
```

```python
x_transformed
```

```python
x_transofrmed=pd.DataFrame(x_transformed,columns=x.columns)
x_transofrmed
```

```python
## feature selection
from sklearn.feature_selection import mutual_info_classif
mf=mutual_info_classif(x_transformed,y) 
```

```python
x.drop(["weight_g","firmness_kgf","tree_age_years","soil_nitrogen_ppm",],axis=1,inplace=True)
```

```python
x.drop(["altitude_m"],axis=1,inplace=True)
```

```python
x.shape
```

```python
# train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
x.info()
```

```python
ct=ColumnTransformer(transformers=[("oe",OrdinalEncoder(),[0,1,4]),
                                   ("sc",StandardScaler(),[2,3,5,6,7])])
```

```python
x_train_transform=ct.fit_transform(x_train)
x_tarin_transorm=pd.DataFrame(x_train_transform)
```

```python
x_test_transform=ct.transform(x_test)
x_test_transform=pd.DataFrame(x_test_transform)
```

```python
## model building
from sklearn.neighbors import KNeighborsClassifier
kn= KNeighborsClassifier(n_neighbors=5,metric="minkowski",p=4)
```

```python
# training
kn.fit(x_train_transform,y_train)
```

```python
# test
y_pred=kn.predict(x_test_transform)
```

```python
# evolution
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
ac=accuracy_score(y_test,y_pred)
ac
```

```python
pr=precision_score(y_test,y_pred,average="micro")
pr
```

```python
f1=f1_score(y_test,y_pred,average="micro")
f1
```

```python
from sklearn.metrics import confusion_matrix,classification_report
cf=confusion_matrix(y_test,y_pred)
cf
```

```python
cr=classification_report(y_test,y_pred)
print(cr)
```
