## label encoder

## One Hot Encoder
- pandas

```python
df=pd.DataFrame({"animals":["cat","Dog","fox"]})
print(df)
```

- sklearn

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(sparse_output=False)
dff=ohe.fit_transform(df)
print(dff)
```

```python
df.job_type.unique()
```

```python
en=pd.get_dummies(df["job_type"],prefix='job_type')
en
```

```python
df=pd.concat([df,en],axis=1)
df
```

```python
encode=ohe.fit_transform(df[['job_type']])
encode
```

```python
enc_df=pd.DataFrame(encode)
```

```python
df=pd.concat([df,enc_df],axis=1)
df
```

## Ordinal encoder

```python
from sklearn.preprocessing import OrdinalEncoder
education=[['ssc'],['ug'],['pg'],['inter']]
oe=OrdinalEncoder(categories=[['ssc','inter','ug','pg']])
ev=oe.fit_transform(education)
ev
```

```python
df.social_platform_preference.unique()
```

```python
df.social_platform_preference.unique()
oe=OrdinalEncoder(categories=[['Facebook', 'Twitter', 'Telegram', 'TikTok', 'Instagram']])
ev=oe.fit_transform(df[['social_platform_preference']])
ev
```

```python
enc_df=pd.DataFrame(ev)
```

```python
df=pd.concat([df,enc_df],axis=1)
df
```

## Column Trnsformer

```python
df = pd.read_csv("social_media.csv")
```

```python
num_col=df.select_dtypes(include="number").columns
print(num_col)
```

```python
cat_col=df.select_dtypes(include="object").columns
print(cat_col)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
pre=ColumnTransformer(transformers=[('ohe',OneHotEncoder(),cat_col),('sc',StandardScaler(),num_col)])
```

```python
pre
```

```python
prep=pre.fit_transform(df)
prep
```

```python
dfff=pd.DataFrame(prep)
dfff
```
