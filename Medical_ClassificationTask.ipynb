## TASK 2: Clean & Preprocess the Text

```python
df["a"] = df["a"].str.lower()
```

#### Removing Punctions

```python
import re
df["a"] = df["a"].apply(
    lambda x: re.sub(r'[^\w\s]', '', str(x))
)
```

#### Removing numbers


```python
df["a"] = df["a"].apply(
    lambda x: re.sub(r'\d+', '', str(x))
)
```

#### Removing extra spaces


```python
df["a"] = df["a"].apply(
    lambda x: re.sub(r'\s+', ' ', str(x)).strip()
)
```

```python
df["a"]
```

#### Remove stopwords

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words=set(stopwords.words('english'))
```

```python
df["a"] = df["a"].apply(
    lambda x: ' '.join(
        word for word in str(x).split()
        if word.lower() not in stop_words
    )
)
```

#### Remove Lemmatize or stem the words

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Download once
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

df["a"] = df["a"].apply(
    lambda x: ' '.join(
        lemmatizer.lemmatize(word) for word in str(x).split()
    )
)

df["a"].head()
```

```python
def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags
        "]+",
        flags=re.UNICODE
    )
    
    return emoji_pattern.sub(r'', str(text))

df["a"] = df["a"].apply(remove_emojis)
```

## TASK 3: Convert Text to Numerical Features

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
    stop_words='english',
    max_features=5000
)

x = cv.fit_transform(df["a"])
```

```python
bow_df = pd.DataFrame(
    x.toarray(),
    columns=cv.get_feature_names_out()
)

bow_df
```

## TASK 4: Split the Data

```python
df.drop(columns=["Unnamed: 0"], inplace=True)
```

```python
X = df["a"]      # text data
y = df["0"]      # target labels
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

## TASK 5: Train Naïve Bayes Models


```python
from sklearn.naive_bayes import MultinomialNB

```

```python
model = MultinomialNB()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy Score:")
print(accuracy_score(y_test, y_pred))

```

```python
from sklearn.metrics import classification_report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix


print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(x_train, y_train)
```

```python
y_pred = knn.predict(x_test)
```
